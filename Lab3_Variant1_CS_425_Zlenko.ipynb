## Завдання 1
1. Функція для підрахунку всіх голосних літер у заданому рядку.
2. Функція, яка приймає число `n` і повертає словник із `n` елементів `{'i': i}`, де `i ∈ [1, n]`.

```python
UA_VOWELS = "аеєиіїоуюя"
EN_VOWELS = "aeiou"
VOWELS = set(UA_VOWELS + EN_VOWELS)

def count_vowels(s):
    return sum(1 for ch in s.lower() if ch in VOWELS)

assert count_vowels("Python") == 1
assert count_vowels("Hello, World!") == 3
assert count_vowels("Привіт, світе!") == 4
assert count_vowels("") == 0
assert count_vowels("рррр ЗЗЗ bcd") == 0
assert count_vowels("AEIOU аеєиіїоуюя") == 15
print("Завдання 1.1: виконано")
```

```python
def create_dict(n):
    return {str(i): i for i in range(1, n + 1)}


assert create_dict(0) == {}
assert create_dict(1) == {'1': 1}
assert create_dict(3) == {'1': 1, '2': 2, '3': 3}
assert len(create_dict(10)) == 10
assert create_dict(10)['10'] == 10
print("Завдання 1.2: виконано")

dict1 = create_dict(10)
dict1
```

## Завдання 2
Робота з текстом із файлу `text.txt`:
- 2.1 підрахувати, скільки разів кожне слово зустрічається в тексті (словник `<word> : <кількість>`);
- 2.2 знайти найуживаніше слово;
- 2.3 впорядкувати словник за ключами (за зростанням);
- 2.4 впорядкувати словник за значеннями (за зростанням).


```python
import os

if not os.path.exists("text.txt"):
    print("не знайдено файл text.txt")
else:
    with open("text.txt", encoding="utf-8") as file:
        text = file.read()

    print(text[:300])
```

```python
import re
def clean_text(text):
    text = text.lower()
    text = re.sub(r"[’'`]s\b", "", text)
    text = re.sub(r"[^\w\s-]", " ", text)
    text = re.sub(r"(?<!\w)-|-(?!\w)", " ", text)
    return text.replace("_", " ")

def count_words(text):
    freq = {}
    for word in clean_text(text).split():
        freq[word] = freq.get(word, 0) + 1
    return freq

t = "The cat. The Dog's (bone), the cat!\n"
assert count_words(t) == {"the": 3, "cat": 2, "dog": 1, "bone": 1}
assert count_words("") == {}
assert count_words("A a A") == {"a": 3}
assert count_words("high-level\nhigh-level, level") == {"high-level": 2, "level": 1}
assert count_words("Python's python") == {"python": 2}

word_freq = count_words(text)
for word, n in word_freq.items():
    print(f"{word} : {n}")
```

```python
def most_common_word(freq):
    if not freq:
        return None
    word = max(freq, key=freq.get)
    return word, freq[word]

assert most_common_word({"a": 1, "b": 5, "c": 2}) == ("b", 5)
assert most_common_word({"x": 1}) == ("x", 1)
assert most_common_word({}) is None
assert most_common_word(count_words("to be or not to be to")) == ("to", 3)

word, n = most_common_word(word_freq)
print(word, " :", n)
```

```python
def sort_by_keys(d):
    return dict(sorted(d.items(), key=lambda item: item[0]))

assert list(sort_by_keys({"b": 2, "a": 1, "c": 3}).keys()) == ["a", "b", "c"]
assert sort_by_keys({}) == {}
assert list(sort_by_keys({"z": 1, "y": 1}).items()) == [("y", 1), ("z", 1)]
assert list(sort_by_keys(count_words("b a c a")).items()) == [("a", 2), ("b", 1), ("c", 1)]

for word, n in sort_by_keys(word_freq).items():
    print(f"{word} : {n}")
```

```python
def sort_by_values(d):
    return dict(sorted(d.items(), key=lambda item: item[1]))

assert list(sort_by_values({"a": 3, "b": 1, "c": 2}).values()) == [1, 2, 3]
assert sort_by_values({}) == {}
assert list(sort_by_values({"a": 2, "b": 1}).keys()) == ["b", "a"]
vals = list(sort_by_values(word_freq).values())
assert vals == sorted(vals)

for word, n in sort_by_values(word_freq).items():
    print(f"{word} : {n}")
```
